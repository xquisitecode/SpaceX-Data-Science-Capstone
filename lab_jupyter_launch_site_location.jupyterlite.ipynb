```python
# Task 2: Mark the success/failed launches for each site on the map


from folium.plugins import MarkerCluster

# --- 1) marker_color column: green = success (class 1), red = failure (class 0)
def assign_marker_color(launch_class):
    return 'green' if launch_class == 1 else 'red'

spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# --- 2) Sanity check: records per site + color counts
print("Launch records per site:")
print(spacex_df['Launch Site'].value_counts())
print("\nMarker color counts:")
print(spacex_df['marker_color'].value_counts())

# --- 3) Map + MarkerCluster
site_map = folium.Map(location=[28.5721, -80.6480], zoom_start=4)
marker_cluster = MarkerCluster().add_to(site_map)

# --- 4) One marker per launch record, colored by outcome
for index, record in spacex_df.iterrows():
    outcome = 'Success' if record['class'] == 1 else 'Failure'
    folium.Marker(
        location=[record['Lat'], record['Long']],
        popup=f"{record['Launch Site']} — {outcome}",
        tooltip=outcome,
        icon=folium.Icon(color='white',
                         icon_color=record['marker_color'],
                         icon='info-sign')
    ).add_to(marker_cluster)

# --- 5) Explicit legend so green/red is unmistakable in the screenshot
legend_html = '''
<div style="position: fixed; bottom: 30px; left: 30px; z-index: 9999;
            background-color: white; padding: 8px 12px; border: 2px solid grey;
            border-radius: 6px; font-size: 14px;">
  <b>Launch Outcome</b><br>
  <span style="color: green;">&#9679;</span> Success (class = 1)<br>
  <span style="color: red;">&#9679;</span> Failure (class = 0)
</div>
'''
try:
    site_map.get_root().html.add_child(folium.Element(legend_html))
except Exception as e:
    print("Legend skipped:", e)

site_map
```

```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import sin, cos, sqrt, atan2, radians

# 1. Define the distance function right here so we don't get an error
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0
    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = R * c
    return distance

# 2. Set coordinates for launch site and coastline
launch_site_lat = 28.563197
launch_site_lon = -80.576820
coastline_lat = 28.56367
coastline_lon = -80.57163

# 3. Calculate the distance
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# 4. Create a new map zoomed in on the launch site
site_map = folium.Map(location=[launch_site_lat, launch_site_lon], zoom_start=14)

# 5. Add the distance marker on the coastline
distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
        )
    )
site_map.add_child(distance_marker)

# 6. Draw a line between the launch site and the coastline
coordinates = [[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]]
lines = folium.PolyLine(locations=coordinates, weight=2, color='blue')
site_map.add_child(lines)

# 7. Add a marker for the launch site itself
folium.Marker([launch_site_lat, launch_site_lon], popup='CCAFS SLC 40').add_to(site_map)

# Display the map
site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

Based on the proximity analysis map:

*   **Are launch sites in close proximity to railways?**
    Yes. Launch sites are built extremely close to railways (like the NASA Railroad) because they are the safest and most efficient way to transport massive, heavy rocket components and fuel.
*   **Are launch sites in close proximity to highways?**
    Yes. They are located near major highways (like Samuel C. Phillips Pkwy) to allow for the transport of smaller equipment, personnel, and tourists.
*   **Are launch sites in close proximity to coastline?**
    Yes, incredibly close. Launching directly over the ocean is a strict safety requirement. If a rocket fails or stages fall back to Earth, they must land in the water to avoid civilian casualties.
*   **Do launch sites keep certain distance away from cities?**
    Yes. They are intentionally placed far from major cities to protect the public from catastrophic failures, massive noise pollution, and sonic booms. 

**Explanation of Findings:**
Launch sites are strategically placed at the intersection of logistics and safety. They are heavily connected to transport networks (railways and highways) to move equipment, but positioned on the coastline to ensure rockets fly out over the ocean, safely away from populated cities.
